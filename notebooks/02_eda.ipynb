## 3. Geographic Patterns

```python
plot_barplot(df["state"])
```

We have imported some data about the US state population to see which states tend to have the most incidents per capita. The data is from 2015.

```python
df_state = pd.read_csv("../data/US_State_Populations_2015.csv")
df_state.set_index("state", inplace=True)
df_state_counts = df["state"].value_counts()
df_state_counts.name = "incidents"

df_killed_per_state = df.groupby("state")["n_killed"].sum().sort_values(ascending=False)
df_injured_per_state = (
    df.groupby("state")["n_injured"].sum().sort_values(ascending=False)
)

df_state = pd.merge(
    df_state, df_state_counts, left_index=True, right_index=True, how="inner"
)
df_state = pd.merge(
    df_state,
    df_killed_per_state.rename("killed"),
    left_index=True,
    right_index=True,
    how="inner",
)
df_state = pd.merge(
    df_state,
    df_injured_per_state.rename("injured"),
    left_index=True,
    right_index=True,
    how="inner",
)

df_state["incidents_per_thousand"] = (
    df_state["incidents"] / df_state["population"]
) * 1_000
df_state["killed_per_thousand"] = (df_state["killed"] / df_state["population"]) * 1_000
df_state["injured_per_thousand"] = (
    df_state["injured"] / df_state["population"]
) * 1_000

df_state.sort_values(by="incidents_per_thousand", ascending=False, inplace=True)
```

```python
usa_incidents_per_thousand = (
    df_state["incidents"].sum() / df_state["population"].sum() * 1_000
)
print(f"USA incidents per thousand people: {usa_incidents_per_thousand:.2f}")

usa_killed_per_thousand = (
    df_state["killed"].sum() / df_state["population"].sum() * 1_000
)
print(f"USA killed per thousand people: {usa_killed_per_thousand:.2f}")

usa_injured_per_thousand = (
    df_state["injured"].sum() / df_state["population"].sum() * 1_000
)
print(f"USA injured per thousand people: {usa_injured_per_thousand:.2f}")
```

```python
import plotly.express as px


def plot_state_map(df, metric, title, color_scale="Reds"):
    fig = px.choropleth(
        df,
        locations="code",
        locationmode="USA-states",
        scope="usa",
        color=metric,
        color_continuous_scale=color_scale,
        hover_name="state",
        hover_data=["incidents", "killed", "injured", "population"],
        title=title,
    )
    fig.update_layout(title_x=0.5, margin=dict(l=0, r=0, t=60, b=0))
    return fig


plot_state_map(
    df_state2, "incidents_per_thousand", "Incidents per 1,000 inhabitants"
).show()
plot_state_map(
    df_state2, "killed_per_thousand", "Killings per 1,000 inhabitants", "OrRd"
).show()
plot_state_map(
    df_state2, "injured_per_thousand", "Injuries per 1,000 inhabitants", "YlOrRd"
).show()
```

### 4.3 People per Incident

```python
people = type_parsed.position.value_counts()

people2 = people.copy()
for i in range(len(people) - 2, -1, -1):
    people[i] -= people2[i + 1]
pnp = people.to_numpy()
pnp  # how much people were involved in the incident (both victims and perpetrators)
```

```python
nonzero_mask = pnp > 0
values = pnp[nonzero_mask]
indices = np.where(nonzero_mask)[0] + 1  # original positions as labels

_, ax = plt.subplots(figsize=(10, 8))
ax.barh(indices.astype(str), values)
ax.set_xlabel("Count")
ax.set_ylabel("Number of People Involved")
ax.set_title("Number of People Involved in Incidents")
plt.tight_layout()
plt.show()
```

We can see that in most cases there was only 1 person involved, however there is a shooting were 103 people took part.

## 7. Persisting Processed DataFrames

We save the cleaned and parsed DataFrames as parquet files under `data/` so downstream CRISP-DM phases (data preparation, modeling) can load them directly without re-running this notebook. Parquet preserves dtypes (including the `datetime64` `date` column and the `id` index of `person_df`).

The six files cover everything non-redundant: the cleaned incidents table, the per-person merged table, the per-state aggregate, the two per-gun tables, and the long-form incident characteristics. The intermediate `*_parsed` DataFrames for age, age_group, type, status, gender, and name are not saved separately because they are already merged into `person_df`.


```python
from pathlib import Path

DATA_DIR = Path("../data")
DATA_DIR.mkdir(exist_ok=True)

df.to_parquet(DATA_DIR / "incidents.parquet")
person_df.to_parquet(DATA_DIR / "persons.parquet")
df_state.to_parquet(DATA_DIR / "per_state.parquet")
gun_parsed.to_parquet(DATA_DIR / "guns.parquet")
gun_stolen_parsed.to_parquet(DATA_DIR / "guns_stolen.parquet")
notes_parsed.to_parquet(DATA_DIR / "incident_characteristics.parquet")

for f in sorted(DATA_DIR.glob("*.parquet")):
    print(f"{f.name:<32} {f.stat().st_size / 1024**2:>8.2f} MB")
```
